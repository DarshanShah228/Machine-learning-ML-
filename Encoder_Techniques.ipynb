```python
import pandas as pd
```

```python
import numpy as np
```

```python
df=pd.read_csv(r"customer.csv")
df
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
le.fit_transform(df["purchased"])
```

```python
df["purchased"]=le.fit_transform(df["purchased"])
df
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe=OrdinalEncoder()
```

```python
oe.fit_transform(df[["review","education"]])
```

```python
df[["review","education"]]=oe.fit_transform(df[["review","education"]])
df
```

```python
#one hot encoder
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder(drop='first',sparse=False)
```

```python
encoded=ohe.fit_transform(df[["gender"]])
```

```python
encoded_df=pd.DataFrame(encoded,columns=ohe.get_feature_names(["gender"]))
encoded_df
```

```python
df=pd.concat([df,encoded_df],axis=1)
df
```

```python
df.drop("gender",axis=1,inplace=True)
```
