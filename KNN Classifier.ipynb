```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
x_train=scaler.fit_transform(x_train1)
x_train
```

```python
x_train1=pd.DataFrame(x_train,columns=x_train1.columns)
```

```python
x_train1
```

```python
x_test=scaler.transform(x_test2)
x_test
```

```python
x_test2=pd.DataFrame(x_test,columns=x_test2.columns)
x_test2
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier(n_neighbors=9)
```

```python
knn.fit(x_train1,y_train)
```

```python
y_pred=knn.predict(x_test2)
y_pred
```

```python
y_test.values
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,precision_score,recall_score,f1_score
```

```python
accuracy_score(y_pred,y_test)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
precision_score(y_test,y_pred)
```

```python
recall_score(y_test,y_pred)
```

```python
f1_score()
```

```python
scores=[]
for i in range(1,16):
    knn=KNeighborsClassifier(n_neighbors=i)
    knn.fit(x_train1,y_train)
    y_ped=knn.predict(x_test2)
    scores.append(accuracy_score(y_test,y_ped))
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(range(1,16),scores)
plt.grid(True)
```
