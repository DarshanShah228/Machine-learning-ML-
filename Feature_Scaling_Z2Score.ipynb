```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"D:\D35\Social_Network_Ads(Feature_Scaling).csv")
df
```

```python
x=df.iloc[:,[2,3]]
x
```

```python
y=df.iloc[:,[4]]
y
```

```python
#split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
x_train
```

```python
from sklearn.preprocessing import StandardScaler
scaled=StandardScaler()
x_train_scaler=scaled.fit_transform(x_train)
```

```python
x_train_scaler=pd.DataFrame(x_train_scaler,columns=x_train.columns)
x_train_scaler
```

```python
x_test_scaler=scaled.fit_transform(x_test)
x_test_scaler
```

```python
x_test_scaler=pd.DataFrame(x_test_scaler,columns=x_test.columns)
x_test_scaler
```

```python
np.round(x_train.describe(),1)
```

```python
np.round(x_train_scaler.describe(),1)
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(15,5))

#before
ax1.set_title("Before Scalling")
ax1.scatter(x_train["Age"],x_train["EstimatedSalary"])



#after
ax2.set_title("After Scalling")
ax2.scatter(x_train_scaler["Age"],x_train_scaler["EstimatedSalary"])
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(15,5))

#before
ax1.set_title("Before Scalling")
sns.kdeplot(x_train["Age"],ax=ax1,label="Age")
sns.kdeplot(x_train["EstimatedSalary"],ax=ax1,label="EstimatedSalary")
ax1.legend()

#after
ax2.set_title("After Scalling")
sns.kdeplot(x_train_scaler["Age"],ax=ax2,label="Age")
sns.kdeplot(x_train_scaler["EstimatedSalary"],ax=ax2,label="EstimatedSalary")
ax2.legend()
```
