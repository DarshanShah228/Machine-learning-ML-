```python
import pandas as pd
df=pd.DataFrame([800,1000,1200,1500,2000])
df
```

```python
x=df.iloc[:,[0]]
x
```

```python
#normaization
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
x_scaler=scaler.fit_transform(x)
x_scaler
```

```python
pre=scaler.transform([[1300]])
pre
```

```python
#StandardScaler
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
x_scaled
```

```python
scaler.mean_
```

```python
scaler.scale_
```
